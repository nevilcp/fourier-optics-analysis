```python
import matplotlib.pyplot as plt

from ImageHandler import *
from Filter import *
from scipy import fftpack
from scipy import signal
```

```python
def padKernal(width,length,kernel):
    xL = length - len(kernel)
    xW = width - len(kernel[0])
    wPadLeft = xW//2
    wPadRight = xW - wPadLeft
    lPadUp = xL//2
    lPadDown = xL - lPadUp
    nkernal = np.array(kernel)
    npad = ((lPadUp,lPadDown),(wPadLeft,wPadRight))
    newkernel = np.pad(kernel, pad_width=npad, mode='constant', constant_values=0)
    return newkernel

def show_real(img):
    plt.figure()
    plt.imshow(img.image, cmap='gray')
    plt.title("Image")
    plt.axis('off')
    plt.show()

def show_fft(img):
    psd2D = np.log(np.abs(img.four)**2 + 1)
    plt.figure()
    plt.imshow(psd2D, cmap='gray')
    plt.title("Fourier")
    plt.axis('off')
    plt.show()

def show_real_fft(img):
    show_real(img)
    show_fft(img)
```
